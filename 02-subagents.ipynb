```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from deepagents import create_deep_agent, SubAgent
from langchain_classic.chains.question_answering.map_reduce_prompt import system_template
from langchain_core.messages import HumanMessage

subagents=SubAgent(
    name="DeepAgents 技术架构调研",
    description="非常了解DeepAgents的技术架构，熟悉三个竞品，能够独立调用DeepAgents的技术架构调研任务",
    system_prompt="你是一个非常熟悉DeepAgents的技术架构智能助手,熟悉三个竞品，能够独立调用DeepAgents的技术架构调研任务",

)

load_dotenv()
model=init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
)
agent=create_deep_agent(
   model=model,
   system_prompt="you are  a helpful assistant",
   subagents=[subagents]
)

res=agent.invoke({"messages":[HumanMessage(f"帮我调研一下deepagent的技术架构，对比三个竞品，写一份3000字的分析报告")]})
print(res)
```
